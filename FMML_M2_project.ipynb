````python
##  one hot encoding of categorical features in data
def onehot_encode(df, columns, prefixes):
    df = df.copy()
    for column, prefix in zip(columns, prefixes):
        dummies = pd.get_dummies(df[column], prefix=prefix)
        df = pd.concat([df, dummies], axis=1)
        df = df.drop(column, axis=1)
    return df
````

````python
data = onehot_encode(
    data,
    columns=['Beverage', 'Beverage_prep'],
    prefixes=['bev', 'bevp']
)
````

````python
## to do
# perform pca on the data and plot the explained variace ratio, what is the optimal number of principal components in this case ?
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Assuming 'data' is your Starbucks nutrition DataFrame

# --- Perform PCA and plot explained variance ratio ---
numeric_columns = data.select_dtypes(include=["int64", "float64"]).columns
pca = PCA()
pca.fit(data[numeric_columns])
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel('Number of components')
plt.ylabel('Cumulative explained variance')
plt.title('PCA Explained Variance Ratio')
plt.show()

# --- Visualize principal components ---
n_components = 2  # Choose based on the explained variance plot
pca = PCA(n_components=n_components)
pca.fit(data[numeric_columns])
pca_data = pca.transform(data[numeric_columns])

pca_df = pd.DataFrame(pca_data, columns=[f'PC{i+1}' for i in range(n_components)])
pca_df['Tea'] = data['Tea']  # Add the 'Tea' column for color-coding

sns.scatterplot(x='PC1', y='PC2', hue='Tea', data=pca_df)
plt.title('PCA Visualization')
plt.show()
````
